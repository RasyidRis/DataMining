## Ringkasan Transformasi Fitur
1. Pembuatan Fitur: Fitur baru belanja_per_item memberikan informasi rata-rata harga keranjang belanja per unit barang.
2. Transformasi Log: Meredam rentang nilai belanja yang timpang sehingga model tidak terdistorsi nilai ekstrem.
3. ColumnTransformer: Mengintegrasikan penskalaan numerik dan encoding kategorikal secara simultan dan siap digunakan untuk pelatihan model.
