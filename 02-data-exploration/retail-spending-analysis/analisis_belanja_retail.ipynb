# 02. Data Exploration
Studi Kasus: Analisis Karakteristik & Distribusi Pengeluaran Belanja Pelanggan Retail

Notebook ini mengeksplorasi karakteristik data transaksi, ringkasan statistik deskriptif, dan visualisasi distribusi nilai belanja tahunan pelanggan.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from analisis_belanja_retail import create_customer_spending_dataset

sns.set_theme(style="whitegrid")

# 1. Load data transaksi pelanggan
df = create_customer_spending_dataset()
print(f"Ukuran dataset: {df.shape[0]} baris, {df.shape[1]} kolom")
df.head()
```

### 2. Ringkasan Statistik Deskriptif
Menganalisis nilai rata-rata (mean), standar deviasi (std), nilai tengah (median), serta kuartil data pengeluaran pelanggan.

```python
df.describe().T[["mean", "std", "min", "25%", "50%", "75%", "max"]].round(2)
```

### 3. Visualisasi Distribusi Total Belanja (Histogram & KDE)
Melihat bentuk sebaran frekuensi total pengeluaran tahunan pelanggan.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["total_belanja_juta"], kde=True, color="#2b5c8f", bins=15, edgecolor="w")
plt.title("Distribusi Frekuensi Total Belanja Pelanggan", fontsize=12, pad=12)
plt.xlabel("Total Belanja Tahunan (Juta IDR)")
plt.ylabel("Frekuensi Pelanggan")
plt.tight_layout()
plt.show()
```

## Temuan Utama Analisis Data
1. Sebaran Pengeluaran: Mayoritas pelanggan membelanjakan akumulasi dana tahunan di kisaran 35 hingga 55 juta IDR dengan kurva distribusi yang cenderung simetris.
2. Variabilitas Pelanggan: Nilai rata-rata belanja berada di kisaran 41.8 juta IDR dengan rentang minimal 13.8 juta IDR hingga maksimal 69.5 juta IDR.
3. Karakteristik Data: Distribusi normal moderat ini menunjukkan data transaksi siap untuk tahapan analisis lanjutan maupun pemodelan segmentasi.
